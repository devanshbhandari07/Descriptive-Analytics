```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

file_path = "Car_DataSet.xlsx"

raw = pd.read_excel(file_path,header=None)

local_data = raw.iloc[2:, 1:8].copy()
local_data.columns = ['F', 'Price', 'Registration', 'Fuel', 'Transmission', 'Mileage', 'Engine']

uk_data = raw.iloc[2:, 9:14].copy()
uk_data.columns = ['F_1', 'Price_1', 'Registration_1', 'Transmission_1', 'Mileage_1']
```

```python
plt.figure(figsize=(8,5))
markers = {'Automatic': 'o', 'Manual': 's'}
for t in local_data['Transmission'].dropna().unique():
    subset = local_data[local_data['Transmission'] == t]
    plt.scatter(subset['Mileage'], subset['Price'], label=str(t), marker=markers.get(t, 'o'), alpha=0.7)
plt.xlabel('Mileage')
plt.ylabel('Price (£)')
plt.title('Price vs Mileage (Local dataset)')
plt.legend(title='Transmission')
plt.tight_layout()
plt.show()
```
